```python
trades_raw["trade_date"] = pd.to_datetime(trades_raw['trade_date'])
```

```python
trades_raw.head()
```

```python
trades_raw.info()
```

```python
trades_raw["trade_month"] = trades_raw["trade_date"].dt.month
trades_raw["trade_year"] = trades_raw["trade_date"].dt.year
```

```python
trades_raw
```

```python
trades_raw.info()
```

```python
#case statement in sql
trades_raw["trade_size"] = trades_raw["volume"].apply(lambda x: 'large' if x > 1500 else 'short')
```

```python
trades_raw
```

```python
trades_raw.query('price > 186')
```

```python
trades_raw.isna().sum()
```

```python
trades_raw = trades_raw.sort_values(["instrument", "trade_date"])
trades_raw["price"] = trades_raw.groupby("instrument")["price"].transform(lambda x: x.interpolate())
trades_raw["volume"] = trades_raw["volume"].fillna(trades_raw["volume"].mean())
```

```python
trades_raw.isna().sum()
```

```python
trades_raw
```

```python
trades_raw.groupby(['instrument', 'trade_year', 'trade_month']).agg({
    'volume': 'sum',
    'price': 'mean'
})
```

```python
trades_raw[trades_raw['instrument'].isin(['TSLA', 'AAPL'])].groupby(['instrument', 'trade_year', 'trade_month']).agg({
    'volume': 'sum',
    'price': 'mean'
}).round(2)
```

```python
amzn = trades_raw[trades_raw['instrument'] == 'AMZN'].sort_values('trade_date').copy()
amzn['7day_rolling_avg'] = amzn['price'].rolling(7).mean()

googl = trades_raw[trades_raw['instrument'] == 'GOOGL'].sort_values('trade_date').copy()
googl['7day_rolling_avg'] = googl['price'].rolling(7).mean()

msft = trades_raw[trades_raw['instrument'] == 'MSFT'].sort_values('trade_date').copy()
msft['7day_rolling_avg'] = msft['price'].rolling(7).mean()

tsla = trades_raw[trades_raw['instrument'] == 'TSLA'].sort_values('trade_date').copy()
tsla['7day_rolling_avg'] = tsla['price'].rolling(7).mean()
```
